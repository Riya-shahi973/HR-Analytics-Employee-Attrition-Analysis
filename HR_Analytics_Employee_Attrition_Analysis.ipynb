## Key Findings – HR Analytics Project

1. **Overall Attrition:** Out of 1,470 employees, 237 employees left the company. The overall attrition rate is 16.12%.

2. **Department Analysis:** The Sales department has a higher attrition rate (20.63%) than HR (19.05%) and Research & Development (13.84%).

3. **Overtime Analysis:** Employees who work overtime have a higher attrition rate (30.53%) compared with employees who do not work overtime (10.44%).

4. **Job Role Analysis:** Sales Representatives have the highest attrition rate (39.76%) among the listed job roles.

5. **Tenure Analysis:** Employees who have worked for 0–1 years have a high attrition rate of 34.88%.

### Conclusion

The analysis shows that attrition is higher among employees who work overtime, employees in the Sales department, Sales Representatives, and employees with shorter company tenure. These patterns can help HR teams identify areas where employee retention efforts may be useful.

## HR Recommendations

1. **Reduce Overtime:** Review employee workloads and provide adequate support to employees who frequently work overtime.

2. **Improve Sales Department Retention:** Identify the reasons for higher attrition in the Sales department and improve employee support.

3. **Support Sales Representatives:** Understand their challenges and consider appropriate training, career growth, and recognition programs.

4. **Improve New Employee Retention:** Provide better onboarding, mentoring, and regular feedback during the first year.

5. **Monitor Attrition Regularly:** Use HR dashboards to track attrition rates and identify changes over time.

### Expected Business Impact

These actions may help HR teams identify employee retention issues, improve employee experience, and make more data-driven decisions.
